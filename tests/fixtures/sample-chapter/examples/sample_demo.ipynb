# Sample notebook

Before the first code cell, text may use numbers like 3 or three.

```python
import json


def average(xs):
    # the notebook may comment differently
    return sum(xs) / len(xs)


values = json.load(open("data/values.json"))
mean = average(values)
print(f"mean {mean:.4f}")
print(f"items {len(values)}")
```

## Reading the result

The mean printed above is the average of the stored values, and the count is one per value. The formula is $m = \frac{1}{n}\sum_i x_i$, as in `mean = 2.5`.

```python
assert mean > 1
print(f"summary: mean {mean:.4f} from {len(values)} items")
```
